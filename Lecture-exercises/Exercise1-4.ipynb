```python
# Exercise 1
url = r"https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
df = pd.read_csv(url)

df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'types']
df.index = [f"flower_{x+1}" for x in range(df.shape[0])]
df
```
